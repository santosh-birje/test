```python
from pyspark.sql import SparkSession
from pyspark.sql.types import *
import random
from datetime import datetime, date, timedelta

spark = SparkSession.builder.getOrCreate()


# -----------------------------
# Configuration
# -----------------------------
num_records = 600000

first_names = [
    "Amit", "Priya", "Rahul", "Sneha", "Vikram",
    "Neha", "Arjun", "Kavita", "Rohan", "Meera"
]

last_names = [
    "Sharma", "Patel", "Verma", "Joshi", "Singh",
    "Kumar", "Nair", "Desai", "Gupta", "Shah"
]

departments = [
    "IT", "HR", "Finance", "Marketing",
    "Sales", "Operations", "Support"
]

job_titles = [
    "Software Engineer",
    "QA Engineer",
    "Project Manager",
    "Data Analyst",
    "HR Executive",
    "Sales Executive",
    "Support Engineer"
]

cities = [
    ("Pune", "Maharashtra", "411001"),
    ("Mumbai", "Maharashtra", "400001"),
    ("Bengaluru", "Karnataka", "560001"),
    ("Hyderabad", "Telangana", "500001"),
    ("Delhi", "Delhi", "110001"),
    ("Chennai", "Tamil Nadu", "600001"),
    ("Kochi", "Kerala", "682001")
]

# -----------------------------
# Generate data
# -----------------------------
data = []

for i in range(1, num_records + 1):

    first_name = random.choice(first_names)
    last_name = random.choice(last_names)

    city, state, postal_code = random.choice(cities)

    dob = date(
        random.randint(1975, 2000),
        random.randint(1, 12),
        random.randint(1, 28)
    )

    hire_date = date(
        random.randint(2015, 2025),
        random.randint(1, 12),
        random.randint(1, 28)
    )

    created_at = datetime.now()

    row = (
        i,                                      # 1 EmployeeID
        first_name,                             # 2 FirstName
        last_name,                              # 3 LastName
        f"{first_name.lower()}.{last_name.lower()}{i}@example.com",  # 4 Email
        f"9{random.randint(100000000, 999999999)}",                  # 5 Phone
        dob,                                    # 6 DateOfBirth
        random.choice(["Male", "Female"]),      # 7 Gender
        random.choice(departments),             # 8 Department
        random.choice(job_titles),              # 9 JobTitle
        round(random.uniform(40000, 150000), 2), # 10 Salary
        hire_date,                              # 11 HireDate
        city,                                   # 12 City
        state,                                  # 13 State
        "India",                                # 14 Country
        postal_code,                            # 15 PostalCode
        random.choice(first_names) + " " +
        random.choice(last_names),              # 16 ManagerName
        random.choice(["Full-Time", "Part-Time", "Contract"]), # 17 EmploymentType
        random.choice(["Active", "Inactive"]),  # 18 Status
        round(random.uniform(2.5, 5.0), 1),     # 19 PerformanceRating
        created_at                              # 20 CreatedAt
    )

    data.append(row)

# -----------------------------
# Schema - 20 columns
# -----------------------------
schema = StructType([
    StructField("EmployeeID", IntegerType(), False),
    StructField("FirstName", StringType(), True),
    StructField("LastName", StringType(), True),
    StructField("Email", StringType(), True),
    StructField("Phone", StringType(), True),
    StructField("DateOfBirth", DateType(), True),
    StructField("Gender", StringType(), True),
    StructField("Department", StringType(), True),
    StructField("JobTitle", StringType(), True),
    StructField("Salary", DoubleType(), True),
    StructField("HireDate", DateType(), True),
    StructField("City", StringType(), True),
    StructField("State", StringType(), True),
    StructField("Country", StringType(), True),
    StructField("PostalCode", StringType(), True),
    StructField("ManagerName", StringType(), True),
    StructField("EmploymentType", StringType(), True),
    StructField("Status", StringType(), True),
    StructField("PerformanceRating", DoubleType(), True),
    StructField("CreatedAt", TimestampType(), True)
])

# -----------------------------
# Create DataFrame
# -----------------------------
df = spark.createDataFrame(data, schema)

# -----------------------------
# Create Temporary View
# -----------------------------
df.createOrReplaceTempView("employee")
```
